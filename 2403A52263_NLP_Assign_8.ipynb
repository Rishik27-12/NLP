# word prediction with bigram

```python
def predict_next_word_bigram(word_sequence, bigram_counts, unigram_counts):
    words_in_sequence = word_sequence.lower().split()
    if not words_in_sequence:
        return "Please provide a word sequence."
    last_word = words_in_sequence[-1]

    potential_next_words = {}
    for (w1, w2), count in bigram_counts.items():
        if w1 == last_word:
            potential_next_words[w2] = count

    if not potential_next_words:
        return f"No bigram found starting with '{last_word}'."
    last_word_unigram_count = unigram_counts.get(last_word, 0)
    if last_word_unigram_count == 0:
        return f"'{last_word}' not found in unigram counts. Cannot predict next word."

    predicted_word = None
    max_probability = -1

    for next_word, bigram_count in potential_next_words.items():
        probability = bigram_count / last_word_unigram_count
        print("probability of  " f'{next_word}' " is ", probability)
        if probability > max_probability:
            max_probability = probability
            predicted_word = next_word

    return predicted_word

```

# word prediction with trigram

```python
def predict_next_word_trigram(word_sequence, trigram_counts, bigram_counts):
    words = word_sequence.lower().split()

    if len(words) < 2:
        return "Please provide at least two words for trigram prediction."

    w1, w2 = words[-2], words[-1]

    potential_next_words = {}
    for (a, b, c), count in trigram_counts.items():
        if a == w1 and b == w2:
            potential_next_words[c] = count

    if not potential_next_words:
        return f"No trigram found starting with '{w1} {w2}'."

    bigram_count = bigram_counts.get((w1, w2), 0)
    if bigram_count == 0:
        return f"Bigram ('{w1}', '{w2}') not found."

    predicted_word = None
    max_probability = -1

    print(f"\nTrigrams starting with '{w1} {w2}':")

    for next_word, trigram_count in potential_next_words.items():
        probability = trigram_count / bigram_count
        print(f"Probability of '{next_word}' is {probability:.4f}")

        if probability > max_probability:
            max_probability = probability
            predicted_word = next_word

    return predicted_word
```

# **Next Word Prediction Using Bi-Gram Counts with Laplace Smoothening**

```python
def predict_next_word_bigram_Laplace(word_sequence, bigram_counts, unigram_counts):
    # Tokenize the input sequence and get the last word
    words_in_sequence = word_sequence.lower().split()
    if not words_in_sequence:
        return "Please provide a word sequence."
    last_word = words_in_sequence[-1]

    # Find potential next words based on bigrams starting with last_word
    potential_next_words = {}
    for (w1, w2), count in bigram_counts.items():
        if w1 == last_word:
            potential_next_words[w2] = count

    if not potential_next_words:
        return f"No bigram found starting with '{last_word}'."

    # Calculate probabilities for potential next words
    # P(w2 | w1) = Count(w1, w2) / Count(w1)
    last_word_unigram_count = unigram_counts.get(last_word, 0)
    if last_word_unigram_count == 0:
        return f"'{last_word}' not found in unigram counts. Cannot predict next word."

    predicted_word = None
    max_probability = -1

    for next_word, bigram_count in potential_next_words.items():
        probability = (bigram_count+1) / (last_word_unigram_count+V)
        print("probability of " f'{next_word}' " is ", probability)
        if probability > max_probability:
            max_probability = probability
            predicted_word = next_word

    return predicted_word

```

# **Next Word Prediction Using Tri-Gram Counts based on laplace smoothening**

```python
import collections

# Corpus Definitions
D1_corpus = "I am a student at SR University"
D2_corpus = "I am a professional cricket player"
D3_corpus = "I am studying Btech in CSE AIML"
D4_corpus = "I am currently learning NLP"
D5_corpus = "I have completed my ML project"
D6_corpus = "I am intrested in sports"
D7_corpus = "I was a state cricket team captain"
D8_corpus = "I am planning to do my masters in USA"
D9_corpus = "I Will do my assignments daily"
D10_corpus = "I am buiding aproject in NLP"

# Combined text for unigram, bigram, and trigram (using .lower() for consistency)
combined_text = f"{D1_corpus} {D2_corpus} {D3_corpus} {D4_corpus} {D5_corpus} {D6_corpus} {D7_corpus} {D8_corpus} {D9_corpus} {D10_corpus}".lower()
words = combined_text.split()

# Unigram counts
word_counts = collections.Counter(words)
V = len(word_counts)

# Bigram counts
bigram_list = []
for i in range(len(words) - 1):
    bigram_list.append((words[i], words[i+1]))
bigram_counts = collections.Counter(bigram_list)

# Trigram counts
trigram_list = []
for i in range(len(words) - 2):
    trigram_list.append((words[i], words[i+1], words[i+2]))
trigram_counts = collections.Counter(trigram_list)

def predict_next_word_trigram_Laplace(word_sequence, Trigrams_counts, bigram_counts):
    # Tokenize the input sequence
    words_in_sequence = word_sequence.lower().split()
    if not words_in_sequence:
        return "Please provide a word sequence."

    # Ensure at least two words for trigram prediction
    if len(words_in_sequence) < 2:
        return "Sequence must contain at least two words for trigram prediction."

    # Get the last two words as a tuple
    last_two_words_tuple = tuple(words_in_sequence[-2:])

    # Find potential next words based on trigrams starting with the last two words
    potential_next_words = {}
    for (w1, w2, w3), count in Trigrams_counts.items():
        if (w1, w2) == last_two_words_tuple:
            potential_next_words[w3] = count

    if not potential_next_words:
        return f"No trigram found starting with '{' '.join(last_two_words_tuple)}."

    # Calculate probabilities for potential next words
    # P(w3 | w1,w2) = Count(w1, w2, w3) / Count(w1, w2)
    # The denominator should be the count of the bigram (w1, w2)
    last_two_words_bigram_count = bigram_counts.get(last_two_words_tuple, 0)
    if last_two_words_bigram_count == 0:
        return f"'{' '.join(last_two_words_tuple)}' not found as a bigram. Cannot predict next word."

    predicted_word = None
    max_probability = -1

    for next_word, trigram_count in potential_next_words.items():
        probability = (trigram_count+1) / (last_two_words_bigram_count+V)
        print("probability of  " f'{next_word}' " is ", probability)
        if probability > max_probability:
            max_probability = probability
            predicted_word = next_word

    return predicted_word

# Example usage:
# Try with a word sequence
squence1 = "I am"
next_word1 = predict_next_word_bigram_Laplace(sequence1, bigram_counts, word_counts)
print(f"Given sequence: '{sequence1}', predicted next word: '{next_word1}'")

squence2 = "my"
next_word2 = predict_next_word_bigram_Laplace(squence2, bigram_counts, word_counts)
print(f"Given sequence: '{squence2}', predicted next word: '{next_word2}'")

squence3 = "in"
next_word3 = predict_next_word_bigram_Laplace(squence3, bigram_counts, word_counts)
print(f"Given sequence: '{squence3}', predicted next word: '{next_word3}'")

squence4 = "cricket"
next_word4 = predict_next_word_bigram_Laplace(squence4, bigram_counts, word_counts)
print(f"Given sequence: '{squence4}', predicted next word: '{next_word4}'")
```
